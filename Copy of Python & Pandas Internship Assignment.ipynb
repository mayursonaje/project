Part 5 – Bonus Task


new column net amount

```python
df['net_amount'] = df['final_amount'] - df['Discount']
df.head()
```

```python
def get_customer_report(customer_name):
    data = df[df["Customer_Name"] == customer_name]
    top_product = data.groupby("Product")["Quantity"].sum().idxmax()
    return {
        "Customer Name": customer_name,
        "Total Orders": len(data),
        "Total Quantity Purchased": int(data["Quantity"].sum()),
        "Total Purchase Amount": float(round(data["final_amount"].sum(), 2)),
        "Most Purchased Product": top_product,
    }

def get_city_report(city_name):
    data = df[df["City"] == city_name]
    top_product = data.groupby("Product")["Quantity"].sum().idxmax()
    return {
        "City": city_name,
        "Total Customers": data["Customer_Name"].nunique(),
        "Total Orders": len(data),
        "Total Quantity Sold": int(data["Quantity"].sum()),
        "Total Sales": float(round(data["final_amount"].sum(), 2)),
        "Top-Selling Product": top_product,
    }
```

```python

print(get_customer_report('Priya Patel'))
```

```python
print(get_city_report('Jaipur'))
```

```python
pd.DataFrame([get_city_report(c) for c in df["City"].unique()]).to_csv("city_sales_report.csv", index=False)
pd.DataFrame([get_customer_report(c) for c in df["Customer_Name"].unique()]).to_csv("customer_sales_report.csv", index=False)
df.groupby("Product")[["Quantity", "final_amount"]].sum().to_csv("product_sales_report.csv")
```
